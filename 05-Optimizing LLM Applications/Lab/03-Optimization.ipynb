## Batch Job Inline

```python

from google import genai
from google.genai import types

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

# A list of dictionaries, where each is a GenerateContentRequest
inline_requests = [
    {
        'contents': [{
            'parts': [{'text': 'Tell me a one-sentence joke.'}],
            'role': 'user'
        }]
    },
    {
        'contents': [{
            'parts': [{'text': 'Why is the sky blue?'}],
            'role': 'user'
        }]
    }
]

inline_batch_job = client.batches.create(
    model="models/gemini-3.5-flash-lite",
    src=inline_requests,
    config={
        'display_name': "inlined-requests-job-1",
    },
)

print(f"Created batch job: {inline_batch_job.name}")
```

```python
import time
```

```python
# wait for the job to finish
job_name = inline_batch_job.name
print(f"Polling status for job: {job_name}")

while True:
    batch_job_inline = client.batches.get(name=job_name)
    if batch_job_inline.state.name in ('JOB_STATE_SUCCEEDED', 'JOB_STATE_FAILED', 'JOB_STATE_CANCELLED', 'JOB_STATE_EXPIRED'):
        break
    print(f"Job not finished. Current state: {batch_job_inline.state.name}. Waiting 30 seconds...")
    time.sleep(30)

print(f"Job finished with state: {batch_job_inline.state.name}")

# print the response
for i, inline_response in enumerate(batch_job_inline.dest.inlined_responses, start=1):
    print(f"\n--- Response {i} ---")

    # Check for a successful response
    if inline_response.response:
        # The .text property is a shortcut to the generated text.
        print(inline_response.response.text)
```

## Batch job File

```python

import json
from google import genai
from google.genai import types

client = genai.Client()

# Create a sample JSONL file
with open("my-batch-requests.jsonl", "w") as f:
    requests = [
        {"key": "request-1", "request": {"contents": [{"parts": [{"text": "Describe the process of photosynthesis."}]}]}},
        {"key": "request-2", "request": {"contents": [{"parts": [{"text": "What are the main ingredients in a Margherita pizza?"}]}]}}
    ]
    for req in requests:
        f.write(json.dumps(req) + "\n")

# Upload the file to the File API
uploaded_file = client.files.upload(
    file='my-batch-requests.jsonl',
    config=types.UploadFileConfig(display_name='my-batch-requests', mime_type='jsonl')
)

print(f"Uploaded file: {uploaded_file.name}")
```

```python
from google import genai

# Assumes `uploaded_file` is the file object from the previous step
client = genai.Client()
file_batch_job = client.batches.create(
    model="gemini-2.5-flash",
    src=uploaded_file.name,
    config={
        'display_name': "file-upload-job-1",
    },
)

print(f"Created batch job: {file_batch_job.name}")
```

```python
import json
from google import genai

client = genai.Client()

# Use the name of the job you want to check
# e.g., inline_batch_job.name from the previous step
while True:
    job_name = file_batch_job.name
    batch_job = client.batches.get(name=job_name)

    if batch_job.state.name == 'JOB_STATE_SUCCEEDED':

        # If batch job was created with a file
        if batch_job.dest and batch_job.dest.file_name:
            # Results are in a file
            result_file_name = batch_job.dest.file_name
            print(f"Results are in file: {result_file_name}")

            print("Downloading result file content...")
            file_content = client.files.download(file=result_file_name)
            # Process file_content (bytes) as needed
            print(file_content.decode('utf-8'))

        # If batch job was created with inline request
        # (for embeddings, use batch_job.dest.inlined_embed_content_responses)
        elif batch_job.dest and batch_job.dest.inlined_responses:
            # Results are inline
            print("Results are inline:")
            for i, inline_response in enumerate(batch_job.dest.inlined_responses):
                print(f"Response {i+1}:")
                if inline_response.response:
                    # Accessing response, structure may vary.
                    try:
                        print(inline_response.response.text)
                    except AttributeError:
                        print(inline_response.response) # Fallback
                elif inline_response.error:
                    print(f"Error: {inline_response.error}")
        else:
            print("No results found (neither file nor inline).")
        break
    else:
        print(f"Job did not succeed. Final state: {batch_job.state.name}")
        if batch_job.error:
            print(f"Error: {batch_job.error}")
    time.sleep(30)
```

## Parallel Execution

```python
import asyncio

async def call_model(prompt):
    try:
        # run the blocking model.invoke in a thread so asyncio can run many in parallel
        resp = await asyncio.to_thread(model.invoke, prompt)
        return resp
    except Exception as e:
        return e

async def main():
    prompts = [f"Provide a one-sentence summary about AI topic #{i+1}." for i in range(10)]
    tasks = [asyncio.create_task(call_model(p)) for p in prompts]
    results = await asyncio.gather(*tasks)

    for i, r in enumerate(results, 1):
        print(f"\n--- Result {i} ---")
        if isinstance(r, Exception):
            print("Error:", r)
            continue
        # print common response attributes safely
        if hasattr(r, "content"):
            print(r.content)
        elif hasattr(r, "text"):
            print(r.text)
        else:
            print(r)

# run
await main()
```

## LLMLite

```python

class LLMLiteProxy:
    def __init__(self, lite_model, full_model):
        self.lite_model = lite_model
        self.full_model = full_model

    def invoke(self, prompt, use_lite=True, **kwargs):
        if use_lite:
            return self.lite_model.invoke(prompt, **kwargs)
        else:
            return self.full_model.invoke(prompt, **kwargs)

# Instantiate both models (assuming you have the same API key and setup)
lite_model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", api_key=os.getenv("GEMINI_API_KEY"))
full_model = ChatGoogleGenerativeAI(model="gemini-3.7-flash", api_key=os.getenv("GEMINI_API_KEY"))

# Create the proxy
proxy = LLMLiteProxy(lite_model, full_model)

# Example usage:
prompt = "Summarize the theory of relativity in one sentence."
lite_response = proxy.invoke(prompt, use_lite=True)
full_response = proxy.invoke(prompt, use_lite=False)

print("Lite Model:", lite_response.content)
print("Full Model:", full_response.content)
```
